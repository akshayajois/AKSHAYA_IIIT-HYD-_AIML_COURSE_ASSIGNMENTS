5.What will happen if we take more that two classes at a time?

Answer:

When more than two classes are considered the problem becomes a multi-class classification problem

For example:

0 vs 1 → Binary Classification

0 vs 1 vs 2 vs 3 → Multi-class Classification

The feature distributions may overlap between different classes making classification more difficult. Additional features or a suitable classifier may be required
